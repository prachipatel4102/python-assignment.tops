```python
#q1
def print_playlist_song(songs):
    if not songs:
        return

        print(songs[0])
        print_playlist_song(songs[1:])

playlist=["Shape of you","i wanna be yours","nadaniya","bairan"]
print_playlist_song(playlist)

#Q2
def count_unread_messages(group):
    total=group.get("count",0)

    for subgroup in group.get("subgroups",[]):
        total+=count_unread_messages(subgroup)
    return total

messages={
    "name": "WhatsApp",
    "count": 5,
    "subgroups": [
        {
            "name": "Family",
            "count": 3,
            "subgroups": []
        },
        {
            "name": "Friends",
            "count": 2,
            "subgroups": [
                {
                    "name": "College",
                    "count": 4,
                    "subgroups": []
                }
            ]
        }
    ]
}
print("total unread messages",count_unread_messages(messages))

#Q3
x = "global"
def outer():
    x = "outer"

    def inner():
        nonlocal x
        x = "inner"

    inner()
    print("Inside outer:", x)

outer()
print("Outside:", x)

#Q4
def format_number_short(n):
    if n < 1000:
        return str(n)
    elif n < 1000000:
        return str(round(n / 1000, 1)) + "K"
    elif n < 1000000000:
        return str(round(n / 1000000, 1)) + "M"
    else:
        return str(round(n / 1000000000, 1)) + "B"
print(format_number_short(500))
print(format_number_short(1500))
print(format_number_short(1200000))
print(format_number_short(2500000000))

```
