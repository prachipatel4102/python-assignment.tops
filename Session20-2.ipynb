```python
#Q1
import requests
url="https://jsonplaceholder.typicode.com/posts"
response=requests.get(url)
print("status code:",response.status_code)
posts=response.json()
print("first post title",posts[0]["title"])

#Q2
import requests
url="https://jsonplaceholder.typicode.com/posts"
data={
    "title":"my first post",
    "body":"hello from python",
    "userid":101
}
response=requests.post(url,json=data)
print("status code",response.status_code)
print("response json")
print(response.json())

#Q3
import requests
url="https://jsonplaceholder.typicode.com/users"
response=requests.get(url)
users=response.json()
print("user with .org email")
for user in users:
    if user["email"].endswith(".org"):
        print(user["username"])

#Q4
import requests
url= "http://www.omdbapi.com/"
params={
    "apikey":"demo",
    "s":"avengers"
}
response=requests.get(url,params=params)
data=response.json
print("response",data)
if "totalresults" in data:
    print("total result found:",data["totalresults"])
else:
    print("error:",data.get("error","no result found"))
```
