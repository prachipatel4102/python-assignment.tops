```python
#Q1
from ssl import ALERT_DESCRIPTION_CERTIFICATE_UNOBTAINABLE
import requests
import json

url = "https://api.binance.com/api/v3/ticker/24hr"
try:
    response = requests.get(url)
    response.raise_for_status()

    data = response.json()
    top_10 = data[:10]

    with open("crypto_data.json", "w") as file:
        json.dump(top_10, file, indent=4)

    print("Data saved successfully to crypto_data.json")

except requests.exceptions.RequestException as e:
    print("Error:", e)

#Q2
import json

def find_most_volatile_coin(data):
    max_coin = max(data, key=lambda coin: abs(float(coin["priceChangePercent"])))
    return max_coin["symbol"]
with open("crypto_data.json", "r") as file:
    data = json.load(file)
print("Most Volatile Coin:", find_most_volatile_coin(data))

#Q3
import json
with open("crypto_data.json", "r") as file:
    data = json.load(file)
prices = [float(coin["lastPrice"]) for coin in data]
average_price = sum(prices) / len(prices)

print("Average Price:", round(average_price, 2))
print("\nCoins Below Average Price:")

for coin in data:
    if float(coin["lastPrice"]) < average_price:
        print(coin["symbol"], "-", coin["lastPrice"])

#Q4

doubt

#Q5

doubt
```
