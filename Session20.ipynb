```python
#Q1
class playlist:
    def __init__(self):
        self._songs=[]
    def  add_song(self,song):
        self._songs.append(song)
    def show_playlist(self):
        print("playlist",self._songs)

playlist=playlist()
playlist.add_song("Shape of you")
playlist.add_song("Levitating")
playlist.add_song("Senorita")
playlist.show_playlist()

#Q2
class product:
    def __init__(self):
        self._price=0
    def set_price(self,price):
        self._price=price
    def get_price(self):
        return self._price

product=product()
product.set_price(49999)
print("product price:",product.get_price())

#Q3
class movie:
    def __init__(self):
        self._rating=0
    def set_rating(self,rating):
        if rating >=0 and rating <=10:
            self._rating=rating

        else:
            print("invalid rating!rating must be between 0 and 10")
    def get_rating(self):
        return self._rating

movie=movie()
movie.set_rating(8.5)
print("Movie rating",movie.get_rating())
movie.set_rating(12)

#Q4 
from abc import ABC,abstractmethod
class paymentmethod(ABC):
    @abstractmethod
    def pay(self,amount):
        pass
class paytm(paymentmethod):
    def pay(self,amount):
        print("paid",amount,"using paytm")
class phonepe(paymentmethod):
    def pay(self,amount):
        print("paid",amount,"using phonepe")
p1=paytm()
p2=phonepe()
p1.pay(500)
p2.pay(1000)
```
