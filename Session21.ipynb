```python
#Q1
class product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def get_discounted_price(self):
        return self.price-(self.price *10/100)

product=product("shoes",2000)
print("product name:",product.name)
print("discounted price:",product.get_discounted_price())

#Q2
class product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def get_discounted_price(self):
        return self.price-(self.price *10/100)
class electronics(product):
    def get_discounted_price(self):
        return self.price-(self.price *20/100)

product=electronics("laptop",15000)
print("product name:",product.name)
print("discounted price",product.get_discounted_price())

#Q3
class product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def get_discounted_price(self):
        return self.price-(self.price *10/100)
class Electronics(product):
    def get_discounted_price(self):
        return self.price-(self.price *20/100)
def show_final_price(item):
        print("product",item.name)
        print("discounted price",item.get_discounted_price())
product=product("shoes",5000)
electronics=electronics("laptop",15000)
show_final_price(product)
show_final_price(electronics)

#Q4
class ticket:
    def __init__(self,price):
        self.price=price
    def get_final_price(self):
        return self.price
class premiumticket(ticket):
    def get_final_price(self):
        return super().get_final_price() + 50

ticket=ticket(300)
premiumticket=premiumticket(300)
print("Normal ticket price",ticket.get_final_price())
print("premium ticket price",premiumticket.get_final_price())
```
