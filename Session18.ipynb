```python
#Q1
import re
text = """
Call me at +91-9876543210 or +91-9123456789.
Invalid numbers: +91-12345, 9876543210
"""
pattern = r"\+91-\d{10}"
numbers = re.findall(pattern, text)
print("Phone Numbers:", numbers)

#q2
import re

def has_date(text):
    pattern = r"\b\d{2}/\d{2}/\d{4}\b"

    if re.search(pattern, text):
        return True
    else:
        return False
print(has_date("Meeting is on 15/08/2026"))   
print(has_date("Meeting tomorrow"))  

#Q3
import re
text = """
Shirt: Rs. 299
Shoes: Rs. 1500
Watch: Rs. 4500
Bag: Rs. 999
"""
prices = re.findall(r"Rs\.\s*(\d+)", text)
prices = [int(price) for price in prices]
print("Prices:", prices)
print("Total:", sum(prices))

#Q4
import re

text = """
Contact us at support@gmail.com
or admin123@yahoo.com
or test_user@company.in
"""
pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}"
result = re.sub(pattern, "[hidden email]", text)
print(result)

#Q5
import re

with open("comments.txt", "r") as file:
    text = file.read()
pattern = r"@[A-Za-z0-9_]{3,}"
usernames = re.findall(pattern, text)
unique_usernames = set(usernames)
print("Unique Usernames:")
for username in sorted(unique_usernames):
    print(username)
```
